```python

# VANISHING AND EXPLODING GRADIENTS
# ============================================================

import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# 1. SIGMOID DERIVATIVE
# ============================================================

def sigmoid(x):
    return 1 / (1 + np.exp(-x))


def sigmoid_derivative(x):
    return x * (1 - x)


# ============================================================
# 2. TANH DERIVATIVE
# ============================================================

def tanh_derivative(x):
    return 1 - np.tanh(x) ** 2


# ============================================================
# 3. RELU DERIVATIVE
# ============================================================

def relu_derivative(x):
    return np.where(x > 0, 1, 0)


# ============================================================
# 4. INPUT VALUES
# ============================================================

x = np.linspace(-5, 5, 200)


# ============================================================
# 5. DERIVATIVES
# ============================================================

sigmoid_grad = sigmoid_derivative(
    sigmoid(x)
)

tanh_grad = tanh_derivative(x)

relu_grad = relu_derivative(x)



```

```python

# 6. PLOT SIGMOID GRADIENT
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    x,
    sigmoid_grad,
    label="Sigmoid Derivative"
)

plt.xlabel("Input")
plt.ylabel("Gradient")
plt.title("Sigmoid Gradient")

plt.legend()

plt.show()



```

```python
# 7. PLOT TANH GRADIENT
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    x,
    tanh_grad,
    label="Tanh Derivative"
)

plt.xlabel("Input")
plt.ylabel("Gradient")
plt.title("Tanh Gradient")

plt.legend()

plt.show()
```

```python
# 8. PLOT RELU GRADIENT
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    x,
    relu_grad,
    label="ReLU Derivative"
)

plt.xlabel("Input")
plt.ylabel("Gradient")
plt.title("ReLU Gradient")

plt.legend()

plt.show()
```

```python
# 9. VANISHING GRADIENT SIMULATION
# ============================================================

small_gradient = 0.5

layers = np.arange(1, 21)

vanishing_values = []

value = 1.0

for layer in layers:

    value = value * small_gradient

    vanishing_values.append(value)


# ============================================================



```

```python
# 10. PLOT VANISHING GRADIENT
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    layers,
    vanishing_values,
    marker="o"
)

plt.xlabel("Number of Layers")
plt.ylabel("Gradient Magnitude")

plt.title("Vanishing Gradient Simulation")

plt.show()
```

```python
# ============================================================
# 11. EXPLODING GRADIENT SIMULATION
# ============================================================

large_gradient = 1.5

exploding_values = []

value = 1.0

for layer in layers:

    value = value * large_gradient

    exploding_values.append(value)


# ============================================================
# 12. PLOT EXPLODING GRADIENT
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    layers,
    exploding_values,
    marker="o"
)

plt.xlabel("Number of Layers")
plt.ylabel("Gradient Magnitude")

plt.title("Exploding Gradient Simulation")

plt.show()


# ============================================================
# 13. GRADIENT CLIPPING
# ============================================================

gradients = np.array([
    0.5,
    1.2,
    3.5,
    8.0,
    15.0,
    25.0
])

max_norm = 5

gradient_norm = np.linalg.norm(gradients)

clipped_gradients = (
    gradients *
    min(1, max_norm / gradient_norm)
)


print("Original Gradients:")
print(gradients)

print("\nGradient Norm:")
print(gradient_norm)

print("\nClipped Gradients:")
print(clipped_gradients)

print("\nClipped Gradient Norm:")
print(np.linalg.norm(clipped_gradients))
```
