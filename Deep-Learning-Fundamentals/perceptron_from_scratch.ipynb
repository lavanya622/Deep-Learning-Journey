# Perceptron From Scratch

This practical demonstrates how a single artificial neuron works.

We implement a perceptron manually using:
- Inputs
- Weights
- Bias
- Weighted Sum
- Step Activation Function
- Prediction

```python
# import libraries
import numpy as np
```

```python
#STEP 2 — Create dataset
X = np.array([
    [1, 1],
    [1, 0],
    [0, 1],
    [0, 0]
])

y = np.array([1, 1, 0, 0])
```

```python
# STEP 3 — Create weights
weights = np.array([0.5, 0.5])
# STEP 4 — Create bias
bias = -0.5
```

```python
# STEP 5 — Weighted Sum
weighted_sum = np.dot(X, weights) + bias
print(weighted_sum)
```

```python
# STEP 6 — Step Activation Function
def step_function(z):
    return np.where(z >= 0, 1, 0)
```

```python
# STEP 7 — Make predictions
y_pred = step_function(weighted_sum)

print(y_pred)
```

```python
# step 8 — Print inputs, weights, bias, weighted sum, predictions, and actual values
print("Inputs:")
print(X)

print("\nWeights:")
print(weights)

print("\nBias:")
print(bias)

print("\nWeighted Sum:")
print(weighted_sum)

print("\nPredictions:")
print(y_pred)

print("\nActual:")
print(y)
```
