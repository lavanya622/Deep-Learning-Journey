```python
# import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
# step 1 — create the dataset
X = np.array([
    [1, 1],
    [1, 0],
    [0, 1],
    [0, 0]
])

y = np.array([1, 1, 0, 0])
```

```python
#step 2 — Initialize weights and bias
weights = np.array([0.0, 0.0])
bias = 0.0
```

```python
# step 3 — Define the learning rate
learning_rate = 0.1
```

```python
# Step 5 — Create Step Function
def step_function(z):
    return 1 if z >= 0 else 0
```

```python
# Step 6 — write the learning algorithm
epochs = 10

for epoch in range(epochs):

    total_error = 0

    for i in range(len(X)):

        # Step 1: Calculate weighted sum
        z = np.dot(X[i], weights) + bias

        # Step 2: Make prediction
        y_pred = step_function(z)

        # Step 3: Calculate error
        error = y[i] - y_pred

        # Step 4: Update weights
        weights = weights + learning_rate * error * X[i]

        # Step 5: Update bias
        bias = bias + learning_rate * error

        # Step 6: Calculate total error
        total_error += abs(error)

    print(
        f"Epoch {epoch + 1}: "
        f"Weights = {weights}, "
        f"Bias = {bias}, "
        f"Error = {total_error}"
    )
```

```python
# Step 8 — Test the trained perceptron



print("Final Weights:", weights)
print("Final Bias:", bias)

for i in range(len(X)):

    z = np.dot(X[i], weights) + bias
    prediction = step_function(z)

    print(
        f"Input: {X[i]} | "
        f"Actual: {y[i]} | "
        f"Predicted: {prediction}"
    )
```

```python
# Step 9 — See the learning more clearly


weights = np.array([0.0, 0.0])
bias = 0.0

learning_rate = 0.1
epochs = 10

errors = []


for epoch in range(epochs):

    total_error = 0

    for i in range(len(X)):

        z = np.dot(X[i], weights) + bias

        y_pred = step_function(z)

        error = y[i] - y_pred

        weights = weights + learning_rate * error * X[i]

        bias = bias + learning_rate * error

        total_error += abs(error)

    errors.append(total_error)

    print(
        f"Epoch {epoch + 1}: "
        f"Error = {total_error}"
    )
```

```python
# Step 10 — Plot the error
import matplotlib.pyplot as plt

plt.plot(range(1, epochs + 1), errors, marker="o")

plt.xlabel("Epoch")
plt.ylabel("Total Error")
plt.title("Perceptron Learning Progress")

plt.show()
```
