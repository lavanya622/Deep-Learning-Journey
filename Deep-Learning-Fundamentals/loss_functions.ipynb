# Loss_Functions

```python

# LOSS FUNCTIONS FROM SCRATCH

import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# 1. MEAN SQUARED ERROR (MSE)

# Mainly used for Regression problems.
#
# Formula:
# MSE = Average of (Actual - Predicted)^2


y_true_reg = np.array([10, 20, 30, 40])

y_pred_reg = np.array([12, 18, 29, 35])


mse = np.mean((y_true_reg - y_pred_reg) ** 2)

print("========== MSE ==========")
print("Actual Values    :", y_true_reg)
print("Predicted Values :", y_pred_reg)
print("MSE              :", mse)


# ============================================================
# 2. BINARY CROSS-ENTROPY (BCE)

# Used for Binary Classification.
#
# Example:
# 0 -> No
# 1 -> Yes
#
# Formula:
# BCE = -mean[
#     y*log(p) + (1-y)*log(1-p)
# ]


def binary_cross_entropy(y_true, y_pred):

    # Prevent log(0)
    epsilon = 1e-15

    y_pred = np.clip(
        y_pred,
        epsilon,
        1 - epsilon
    )

    loss = -np.mean(
        y_true * np.log(y_pred)
        +
        (1 - y_true) * np.log(1 - y_pred)
    )

    return loss


# Good predictions
y_true_binary = np.array([1, 0, 1, 1])

y_pred_binary_good = np.array([
    0.9,
    0.2,
    0.8,
    0.7
])


bce_good = binary_cross_entropy(
    y_true_binary,
    y_pred_binary_good
)


print("\n========== BINARY CROSS-ENTROPY ==========")
print("Actual Values          :", y_true_binary)
print("Good Predictions       :", y_pred_binary_good)
print("Good Prediction Loss   :", bce_good)


# Bad predictions
y_pred_binary_bad = np.array([
    0.1,
    0.8,
    0.2,
    0.3
])


bce_bad = binary_cross_entropy(
    y_true_binary,
    y_pred_binary_bad
)


print("Bad Predictions        :", y_pred_binary_bad)
print("Bad Prediction Loss    :", bce_bad)


# ============================================================
# 3. CATEGORICAL CROSS-ENTROPY (CCE)

# Used for Multi-Class Classification.
#
# Example:
# Cat   -> [1, 0, 0]
# Dog   -> [0, 1, 0]
# Horse -> [0, 0, 1]


def categorical_cross_entropy(y_true, y_pred):

    # Prevent log(0)
    epsilon = 1e-15

    y_pred = np.clip(
        y_pred,
        epsilon,
        1 - epsilon
    )

    loss = -np.sum(
        y_true * np.log(y_pred)
    )

    return loss


# Actual class = Cat
y_true_categorical = np.array([
    1,
    0,
    0
])


# Good prediction
y_pred_categorical_good = np.array([
    0.8,
    0.1,
    0.1
])


cce_good = categorical_cross_entropy(
    y_true_categorical,
    y_pred_categorical_good
)


print("\n========== CATEGORICAL CROSS-ENTROPY ==========")
print("Actual Class          :", y_true_categorical)
print("Good Prediction       :", y_pred_categorical_good)
print("Good Prediction Loss  :", cce_good)


# Bad prediction
y_pred_categorical_bad = np.array([
    0.1,
    0.8,
    0.1
])


cce_bad = categorical_cross_entropy(
    y_true_categorical,
    y_pred_categorical_bad
)


print("Bad Prediction        :", y_pred_categorical_bad)
print("Bad Prediction Loss   :", cce_bad)


```

```python
# 4. MSE LOSS VISUALIZATION
# Actual value = 10
# We check how the loss changes when prediction changes.


actual = 10

predictions = np.linspace(
    0,
    20,
    100
)

mse_values = (
    actual - predictions
) ** 2


plt.figure(figsize=(8, 5))

plt.plot(
    predictions,
    mse_values
)

plt.xlabel("Prediction")
plt.ylabel("Squared Error")
plt.title("MSE Loss")

plt.grid()

plt.show()
```

```python

# 5. BCE LOSS VISUALIZATION
# Actual class = 1
#
# When prediction approaches 1:
# Loss becomes smaller.
#
# When prediction approaches 0:
# Loss becomes larger.


probabilities = np.linspace(
    0.01,
    0.99,
    100
)

bce_values = -np.log(
    probabilities
)


plt.figure(figsize=(8, 5))

plt.plot(
    probabilities,
    bce_values
)

plt.xlabel("Predicted Probability")
plt.ylabel("Binary Cross-Entropy Loss")
plt.title("BCE Loss for Actual Class = 1")

plt.grid()

plt.show()
```

```python

# 6. FINAL COMPARISON

print("\n========== LOSS FUNCTION COMPARISON ==========")

print("MSE                         :", mse)

print("BCE - Good Prediction      :", bce_good)

print("BCE - Bad Prediction       :", bce_bad)

print("CCE - Good Prediction      :", cce_good)

print("CCE - Bad Prediction       :", cce_bad)
```

```python

# 7. BASIC INTERPRETATION

print("\n========== INTERPRETATION ==========")

print("MSE  -> Mainly used for Regression")

print("BCE  -> Used for Binary Classification")

print("CCE  -> Used for Multi-Class Classification")

print("\nLower Loss  -> Better prediction")

print("Higher Loss -> Worse prediction")
```
