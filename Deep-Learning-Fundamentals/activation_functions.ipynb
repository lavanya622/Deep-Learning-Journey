# Activation_Functions

```python
# import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Step 2 — Create input values
x = np.linspace(-10, 10, 100)
```

```python
# Step 3 — Sigmoid Function

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Now calculate:

y_sigmoid = sigmoid(x)

# Check:

print(y_sigmoid)
```

```python
# Step 4 — Plot Sigmoid
plt.plot(x, y_sigmoid)

plt.xlabel("Input")
plt.ylabel("Sigmoid Output")
plt.title("Sigmoid Activation Function")

plt.grid()
plt.show()
```

```python
# Step 5 — Tanh Function

def tanh(x):
    return np.tanh(x)

# Calculate:

y_tanh = tanh(x)

# Print:

print(y_tanh)
```

```python
# Step 6 — Plot Tanh
plt.plot(x, y_tanh)

plt.xlabel("Input")
plt.ylabel("Tanh Output")
plt.title("Tanh Activation Function")

plt.grid()
plt.show()
```

```python
# Step 7 — ReLU


def relu(x):
    return np.maximum(0, x)

# Calculate:

y_relu = relu(x)

# Check:

print(y_relu)
```

```python
# Step 8 — Plot ReLU
plt.plot(x, y_relu)

plt.xlabel("Input")
plt.ylabel("ReLU Output")
plt.title("ReLU Activation Function")

plt.grid()
plt.show()
```

```python
# Step 9 — Leaky ReLU


def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)

# Calculate:

y_leaky_relu = leaky_relu(x)

# Print:

print(y_leaky_relu)
```

```python
# Step 10 — Plot Leaky ReLU
plt.plot(x, y_leaky_relu)

plt.xlabel("Input")
plt.ylabel("Leaky ReLU Output")
plt.title("Leaky ReLU Activation Function")

plt.grid()
plt.show()
```

```python
# Step 11 — Compare all four

plt.plot(x, sigmoid(x), label="Sigmoid")
plt.plot(x, tanh(x), label="Tanh")
plt.plot(x, relu(x), label="ReLU")
plt.plot(x, leaky_relu(x), label="Leaky ReLU")

plt.xlabel("Input")
plt.ylabel("Output")
plt.title("Comparison of Activation Functions")

plt.legend()
plt.grid()
plt.show()
```

```python
# Step 12 — Create a small comparison table


activation_info = {
    "Sigmoid": "Output range: 0 to 1",
    "Tanh": "Output range: -1 to +1",
    "ReLU": "max(0, x)",
    "Leaky ReLU": "x if x > 0, otherwise 0.01*x"
}

for name, information in activation_info.items():
    print(name, "→", information)
```
