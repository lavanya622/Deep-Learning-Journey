# Mlp Forward_Propagation

```python
import numpy as np

# Input
X = np.array([1.0, 2.0])

# Hidden layer weights and bias
W_hidden = np.array([
    [0.5, 0.3],
    [0.2, 0.4]
])

b_hidden = np.array([0.1, 0.2])

# Output layer weights and bias
W_output = np.array([0.6, 0.5])
b_output = 0.1

# Activation functions
def relu(x):
    return np.maximum(0, x)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Forward propagation
z_hidden = np.dot(X, W_hidden) + b_hidden
a_hidden = relu(z_hidden)

z_output = np.dot(a_hidden, W_output) + b_output
y_pred = sigmoid(z_output)

# Display results
print("Input:", X)
print("Hidden weighted sum:", z_hidden)
print("Hidden output:", a_hidden)
print("Output weighted sum:", z_output)
print("Final prediction:", y_pred)
```
