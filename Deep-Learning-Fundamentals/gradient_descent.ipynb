# Gradient_Descent

```python
# GRADIENT DESCENT FROM SCRATCH


import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# 1. CREATE SIMPLE DATA

# We want the model to learn:
#
# y = 2x
#
# So the correct weight should eventually become close to 2.


X = np.array([1, 2, 3, 4, 5], dtype=float)

y = np.array([2, 4, 6, 8, 10], dtype=float)


# ============================================================
# 2. INITIALIZE WEIGHT


weight = 0.0

learning_rate = 0.01

epochs = 100


# Store loss and weight values
loss_history = []
weight_history = []


# ============================================================
# 3. GRADIENT DESCENT
# ============================================================

for epoch in range(epochs):

    # --------------------------------------------------------
    # Forward Pass
    # --------------------------------------------------------
    
    y_pred = weight * X


    # --------------------------------------------------------
    # Calculate MSE Loss
    # --------------------------------------------------------
    
    loss = np.mean(
        (y - y_pred) ** 2
    )


    # --------------------------------------------------------
    # Calculate Gradient
    # --------------------------------------------------------
    #
    # Derivative of MSE with respect to weight:
    #
    # dL/dw = -2/n * sum(X * (y - y_pred))
    #
    
    gradient = (
        -2 / len(X)
        * np.sum(
            X * (y - y_pred)
        )
    )


    # --------------------------------------------------------
    # Update Weight
    # --------------------------------------------------------
    #
    # w = w - learning_rate * gradient
    #
    
    weight = (
        weight
        - learning_rate * gradient
    )


    # Store values
    loss_history.append(loss)

    weight_history.append(weight)


    # Print progress
    if (epoch + 1) % 10 == 0:

        print(
            f"Epoch {epoch + 1}: "
            f"Weight = {weight:.4f}, "
            f"Loss = {loss:.4f}"
        )


# ============================================================
# 4. FINAL RESULTS


print("\n========== FINAL RESULT ==========")

print("Final Weight:", weight)

print("Final Loss:", loss_history[-1])


# ============================================================
# 5. MAKE FINAL PREDICTIONS

y_pred_final = weight * X


print("\nActual Values   :", y)

print("Predicted Values:", y_pred_final)


```

```python
# 6. PLOT LOSS


plt.figure(figsize=(8, 5))

plt.plot(
    range(1, epochs + 1),
    loss_history
)

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.title("Gradient Descent - Loss Reduction")

plt.grid()

plt.show()
```

```python
# 7. PLOT WEIGHT LEARNING


plt.figure(figsize=(8, 5))

plt.plot(
    range(1, epochs + 1),
    weight_history
)

plt.axhline(
    y=2,
    linestyle="--",
    label="Target Weight = 2"
)

plt.xlabel("Epoch")

plt.ylabel("Weight")

plt.title("Weight Learning During Gradient Descent")

plt.legend()

plt.grid()

plt.show()
```

```python
# 8. PREDICTION GRAPH

plt.figure(figsize=(8, 5))

plt.scatter(
    X,
    y,
    label="Actual"
)

plt.plot(
    X,
    y_pred_final,
    label="Predicted"
)

plt.xlabel("X")

plt.ylabel("Y")

plt.title("Actual vs Predicted")

plt.legend()

plt.grid()

plt.show()
```

```python
#step : 9 LEARNING RATE COMPARISON
# ============================================================

learning_rates = [0.001, 0.01, 0.05]

plt.figure(figsize=(8, 5))


for lr in learning_rates:

    weight = 0.0

    losses = []

    for epoch in range(100):

        # Forward pass
        y_pred = weight * X

        # Loss
        loss = np.mean(
            (y - y_pred) ** 2
        )

        # Gradient
        gradient = (
            -2 / len(X)
            * np.sum(
                X * (y - y_pred)
            )
        )

        # Weight update
        weight = (
            weight
            - lr * gradient
        )

        losses.append(loss)

    plt.plot(
        range(1, 101),
        losses,
        label=f"Learning Rate = {lr}"
    )


plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.title("Learning Rate Comparison")

plt.legend()

plt.grid()

plt.show()
```
